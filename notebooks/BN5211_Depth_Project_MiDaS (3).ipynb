```python

from pathlib import Path

def get_pairs(sequence):
    folder = DATA_DIR / sequence
    rgb_files = sorted(folder.rglob("*_color.png"))

    pairs = []

    for rgb_path in rgb_files:
        prefix = rgb_path.name.replace("_color.png", "")

        candidates = [
            rgb_path.with_name(prefix + "_depth.tiff"),
            rgb_path.with_name(prefix + "_depth.tif"),
            rgb_path.with_name(prefix + "_depth.png")
        ]

        depth_path = next(
            (p for p in candidates if p.exists()),
            None
        )

        if depth_path is not None:
            pairs.append((rgb_path, depth_path))

    return pairs

dev_pairs = get_pairs("trans_t1_a")
test_pairs = get_pairs("trans_t2_b")

print("Development pairs:", len(dev_pairs))
print("Test pairs:", len(test_pairs))

assert len(dev_pairs) == 61
assert len(test_pairs) == 103
```

```python

def load_depth(path):
    raw = cv2.imread(
        str(path),
        cv2.IMREAD_UNCHANGED
    )

    if raw is None:
        raise ValueError(f"Cannot read depth: {path}")

    if raw.ndim != 2:
        raise ValueError("Expected single-channel depth")

    if raw.dtype == np.uint16:
        # ONLY if confirmed as 0-65535 -> 0-100 mm
        depth_mm = (
            raw.astype(np.float32) * 100.0 / 65535.0
        )
    else:
        raise ValueError(
            f"Unverified TIFF format: {raw.dtype}"
        )

    return depth_mm
```

```python

import torch
import torch.nn.functional as F
from pathlib import Path

MIDAS_DIR = Path("/content/MiDaS")

assert (MIDAS_DIR / "hubconf.py").exists(), (
    "MiDaS repository not found. Run Cell 2 first."
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

# Load the model from the locally cloned repository
midas = torch.hub.load(
    str(MIDAS_DIR),
    "DPT_Hybrid",
    source="local",
    trust_repo=True
)

midas = midas.to(device)
midas.eval()

# Load transforms from the same repository
transforms = torch.hub.load(
    str(MIDAS_DIR),
    "transforms",
    source="local",
    trust_repo=True
)

transform = transforms.dpt_transform

print("Model: MiDaS DPT-Hybrid")
print("Device:", device)
print("Source:", MIDAS_DIR)
```

```python
@torch.inference_mode()
def predict_midas(rgb_image):
    input_batch = transform(rgb_image).to(device)

    prediction = midas(input_batch)

    prediction = F.interpolate(
        prediction.unsqueeze(1),
        size=rgb_image.shape[:2],
        mode="bicubic",
        align_corners=False
    ).squeeze()

    return prediction.cpu().numpy().astype(np.float32)
```

```python

def load_rgb(path):
    img = cv2.imread(str(path), cv2.IMREAD_COLOR)

    if img is None:
        raise ValueError(f"Cannot read image: {path}")

    return cv2.cvtColor(
        img,
        cv2.COLOR_BGR2RGB
    )
```

```python

from tqdm.auto import tqdm

def generate_predictions(pairs):
    predictions = []
    ground_truths = []

    for rgb_path, depth_path in tqdm(pairs):
        rgb_img = load_rgb(rgb_path)
        gt = load_depth(depth_path)

        pred = predict_midas(rgb_img)

        assert pred.shape == gt.shape

        predictions.append(pred)
        ground_truths.append(gt)

    return predictions, ground_truths

dev_preds, dev_gts = generate_predictions(dev_pairs)

print("Development predictions:", len(dev_preds))
```

```python

import numpy as np
from scipy.optimize import least_squares

rng = np.random.default_rng(42)
X_parts, Y_parts = [], []

for pred, gt in zip(dev_preds, dev_gts):
    valid = (
        np.isfinite(pred) &
        np.isfinite(gt) &
        (gt > 0) &
        (gt < 100)
    )

    x = pred[valid].astype(np.float64)
    y = (1.0 / gt[valid]).astype(np.float64)

    if x.size == 0:
        continue

    idx = rng.choice(
        x.size,
        size=min(5000, x.size),
        replace=False
    )

    X_parts.append(x[idx])
    Y_parts.append(y[idx])

X = np.concatenate(X_parts)
Y = np.concatenate(Y_parts)

# Fit inverse depth = a * prediction + b
# a must be positive; b is free.
result = least_squares(
    lambda p: p[0] * X + p[1] - Y,
    x0=[0.001, 0.01],
    bounds=(
        [1e-12, -np.inf],
        [np.inf, np.inf]
    ),
    loss="soft_l1",
    max_nfev=5000
)

a, b = map(float, result.x)

print("Inverse calibration")
print("Scale a:", a)
print("Intercept b:", b)
print("Optimization successful:", result.success)
print("Cost:", result.cost)
```

```python

import numpy as np

def calibrate_depth(pred, a, b):
    inv = (
        a * np.asarray(pred, dtype=np.float64) + b
    )

    if not np.isfinite(inv).all():
        raise ValueError("Non-finite inverse depth")

    if "MIN_DEPTH_MM" not in globals():
        raise RuntimeError(
            "Run Cell 22 to establish depth limits first."
        )

    inv = np.clip(
        inv,
        1.0 / MAX_DEPTH_MM,
        1.0 / MIN_DEPTH_MM
    )

    return (1.0 / inv).astype(np.float32)
```

```python

import numpy as np

valid_depth_values = []

for gt in dev_gts:
    valid = (
        np.isfinite(gt) &
        (gt > 0) &
        (gt < 100)
    )

    valid_depth_values.append(
        gt[valid].astype(np.float64)
    )

dev_values = np.concatenate(valid_depth_values)

MIN_DEPTH_MM = float(np.min(dev_values))
MAX_DEPTH_MM = 100.0
median_depth = float(np.median(dev_values))

print("Minimum development depth:", MIN_DEPTH_MM)
print("Maximum allowed depth:", MAX_DEPTH_MM)
print("Development median:", median_depth)

def predict_baseline(shape):
    return np.full(
        shape,
        median_depth,
        dtype=np.float32
    )
```

```python

def compute_metrics(gt, pred):
    assert gt.shape == pred.shape

    valid = (
        np.isfinite(gt) &
        (gt > 0) &
        (gt < 100)
    )

    gt_valid = gt[valid].astype(np.float64)
    pred_valid = pred[valid].astype(np.float64)

    if len(gt_valid) == 0:
        raise ValueError("No valid pixels")

    if not np.all(np.isfinite(pred_valid)):
        raise ValueError("Non-finite predictions")

    if np.any(pred_valid <= 0):
        raise ValueError("Nonpositive predictions")

    absrel = np.mean(
        np.abs(gt_valid - pred_valid) / gt_valid
    )

    rmse = np.sqrt(
        np.mean((gt_valid - pred_valid) ** 2)
    )

    ratio = np.maximum(
        gt_valid / pred_valid,
        pred_valid / gt_valid
    )

    delta1 = np.mean(ratio < 1.25)

    return {
        "AbsRel": float(absrel),
        "RMSE_mm": float(rmse),
        "Delta1": float(delta1)
    }
```

```python

import numpy as np
from scipy.optimize import lsq_linear

rng = np.random.default_rng(42)

X_parts = []
Y_parts = []

for pred, gt in zip(dev_preds, dev_gts):
    valid = (
        np.isfinite(pred) &
        np.isfinite(gt) &
        (gt > 0) &
        (gt < 100)
    )

    p = pred[valid].astype(np.float64)
    d = gt[valid].astype(np.float64)

    idx = rng.choice(
        len(p),
        size=min(5000, len(p)),
        replace=False
    )

    X_parts.append(p[idx])
    Y_parts.append(d[idx])

X = np.concatenate(X_parts)
Y = np.concatenate(Y_parts)

# Depth = intercept - slope * MiDaS output
# Constrain slope >= 0 because MiDaS predicts inverse depth
A = np.column_stack([-X, np.ones_like(X)])

fit = lsq_linear(
    A,
    Y,
    bounds=([0, 0], [np.inf, np.inf]),
    max_iter=200
)

slope, intercept = fit.x

print("Slope:", slope)
print("Intercept:", intercept)
print("Fit successful:", fit.success)
```

```python

import numpy as np

def calibrate_linear(pred):
    raw_depth = (
        intercept -
        slope * np.asarray(pred, dtype=np.float64)
    )

    if not np.isfinite(raw_depth).all():
        raise ValueError("Non-finite linear depth")

    return np.clip(
        raw_depth,
        MIN_DEPTH_MM,
        MAX_DEPTH_MM
    ).astype(np.float32)
```

```python

import numpy as np
import pandas as pd

def inspect_depths(depth_maps, name):
    values = []

    for gt in depth_maps:
        valid = (
            np.isfinite(gt) &
            (gt > 0) &
            (gt < 100)
        )
        values.append(gt[valid])

    all_depths = np.concatenate(values)

    print(f"\n{name}")
    print("Frames:", len(depth_maps))
    print("Valid pixels:", len(all_depths))
    print("Minimum:", np.min(all_depths))
    print("Maximum:", np.max(all_depths))
    print("Mean:", np.mean(all_depths))
    print("Median:", np.median(all_depths))
    print("Std:", np.std(all_depths))

    return all_depths

dev_values = inspect_depths(dev_gts, "Development")
test_values = inspect_depths(test_gts, "Testing")
```

```python

def valid_depth_mask(gt):
    return (
        np.isfinite(gt) &
        (gt > 0) &
        (gt < 100)
    )

def evaluate_depth(gt, pred):
    gt = np.asarray(gt, dtype=np.float64)
    pred = np.asarray(pred, dtype=np.float64)

    assert gt.shape == pred.shape

    valid = valid_depth_mask(gt)

    g = gt[valid]
    p = pred[valid]

    if len(g) == 0:
        raise ValueError("No valid pixels")

    if not np.all(np.isfinite(p)) or np.any(p <= 0):
        raise ValueError("Invalid predicted depth")

    absrel = np.mean(np.abs(g - p) / g)
    rmse = np.sqrt(np.mean((g - p) ** 2))

    delta = np.mean(
        np.maximum(g / p, p / g) < 1.25
    )

    mae = np.mean(np.abs(g - p))

    return {
        "MAE_mm": float(mae),
        "AbsRel": float(absrel),
        "RMSE_mm": float(rmse),
        "Delta1": float(delta)
    }
```

```python

median_depth_shared = np.median(dev_values)

baseline_results = []

for gt in test_gts:
    pred = np.full_like(
        gt,
        median_depth_shared,
        dtype=np.float32
    )

    baseline_results.append(evaluate_depth(gt, pred))

baseline_df = pd.DataFrame(baseline_results)

print("Development median:", median_depth_shared)
print("\nConstant-median baseline:")
print(baseline_df.mean().round(4))
```

```python

inverse_results = []
linear_results = []

for pred, gt in zip(test_preds, test_gts):
    inverse_depth = calibrate_depth(pred, a, b)
    linear_depth = calibrate_linear(pred)

    inverse_results.append(
        evaluate_depth(gt, inverse_depth)
    )

    linear_results.append(
        evaluate_depth(gt, linear_depth)
    )

inverse_df = pd.DataFrame(inverse_results)
linear_df = pd.DataFrame(linear_results)

comparison = pd.DataFrame({
    "Constant median": baseline_df.mean(),
    "MiDaS inverse calibration": inverse_df.mean(),
    "MiDaS linear calibration": linear_df.mean()
}).T

print(comparison.round(4))

comparison.to_csv(
    RESULTS_DIR / "midas_standardized_comparison.csv"
)
```

```python

def clipping_diagnostics(predictions, ground_truths):
    rows = []

    for i, (pred, gt) in enumerate(
        zip(predictions, ground_truths)
    ):
        raw_depth = intercept - slope * pred
        valid = valid_depth_mask(gt)

        clipped = raw_depth[valid] < 0.1

        rows.append({
            "Frame": i,
            "Clipped_fraction": clipped.mean(),
            "Raw_min_mm": raw_depth[valid].min(),
            "Raw_median_mm": np.median(raw_depth[valid]),
            "Raw_max_mm": raw_depth[valid].max()
        })

    return pd.DataFrame(rows)

clip_dev = clipping_diagnostics(dev_preds, dev_gts)
clip_test = clipping_diagnostics(test_preds, test_gts)

print("Development clipping:", clip_dev["Clipped_fraction"].mean())
print("Test clipping:", clip_test["Clipped_fraction"].mean())

print("\nWorst test frames:")
print(
    clip_test.sort_values(
        "Clipped_fraction", ascending=False
    ).head(10)
)
```

```python

import numpy as np
import pandas as pd
import cv2
from pathlib import Path

def shared_mask(depth_path, rgb_path):
    raw = cv2.imread(
        str(depth_path),
        cv2.IMREAD_UNCHANGED
    )
    rgb = cv2.imread(
        str(rgb_path),
        cv2.IMREAD_COLOR
    )

    if raw is None or rgb is None:
        raise FileNotFoundError(
            f"Missing RGB/depth: {rgb_path}, {depth_path}"
        )

    return (
        (raw > 0) &
        (raw < 65535) &
        (rgb.max(axis=2) > 10)
    )

def shared_metrics(gt, pred, mask):
    g = gt[mask].astype(np.float64)
    p = pred[mask].astype(np.float64)

    if g.size == 0:
        raise ValueError("No valid pixels")

    if not np.isfinite(p).all():
        raise ValueError("Non-finite predictions")

    p = np.clip(p, 1e-3, 100.0)

    err = p - g

    return {
        "MAE_mm": float(np.mean(np.abs(err))),
        "AbsRel": float(np.mean(np.abs(err) / g)),
        "MSE_mm2": float(np.mean(err ** 2)),
        "Delta1": float(np.mean(
            np.maximum(p / g, g / p) < 1.25
        ))
    }

rows = []
clip_rows = []

for i, ((rgb_path, depth_path), pred, gt) in enumerate(
    zip(test_pairs, test_preds, test_gts)
):
    mask = shared_mask(depth_path, rgb_path)

    raw_linear = intercept - slope * pred
    raw_inverse = a * pred + b

    clip_rows.append({
        "frame": rgb_path.stem,
        "linear_below_min_fraction": float(
            np.mean(raw_linear[mask] < MIN_DEPTH_MM)
        ),
        "inverse_outside_bounds_fraction": float(
            np.mean(
                (raw_inverse[mask] < 1 / MAX_DEPTH_MM) |
                (raw_inverse[mask] > 1 / MIN_DEPTH_MM)
            )
        )
    })

    predictions = {
        "Constant median": predict_baseline(gt.shape),
        "MiDaS inverse": calibrate_depth(pred, a, b),
        "MiDaS global linear": calibrate_linear(pred)
    }

    for name, depth in predictions.items():
        rows.append({
            "Frame": rgb_path.stem,
            "Method": name,
            **shared_metrics(gt, depth, mask)
        })

midas_frame_A = pd.DataFrame(rows)

summary_rows = []

for name, group in midas_frame_A.groupby("Method"):
    summary_rows.append({
        "Method": name,
        "MAE_mm": group["MAE_mm"].mean(),
        "AbsRel": group["AbsRel"].mean(),
        "RMSE_mm": np.sqrt(group["MSE_mm2"].mean()),
        "Delta1": group["Delta1"].mean(),
        "n_frames": len(group),
        "Mask": "A"
    })

midas_summary_A = pd.DataFrame(summary_rows)

out = RESULTS_DIR / "standardized_mask_A"
out.mkdir(parents=True, exist_ok=True)

midas_frame_A.to_csv(
    out / "midas_per_frame_mask_A.csv",
    index=False
)

midas_summary_A.to_csv(
    out / "midas_summary_mask_A.csv",
    index=False
)

pd.DataFrame(clip_rows).to_csv(
    out / "midas_clipping_diagnostics.csv",
    index=False
)

print(midas_summary_A.round(4).to_string(index=False))
print("\nSaved to:", out)
```
